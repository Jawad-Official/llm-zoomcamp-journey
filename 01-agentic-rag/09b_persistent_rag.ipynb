# Lesson 9 (part B): Data ingestion: use the saved database

**Simple version:** No downloading this time! We just open `faq.db` and plug it into the same `RAGBase` robot. Same robot, new memory that survives restarts.

Open the saved search database `faq.db` (made in part A).

```python
from sqlitesearch import TextSearchIndex

sqlite_index = TextSearchIndex(
    text_fields=["question", "section", "answer"],
    keyword_fields=["course"],
    db_path="faq.db"
)
```

Load the API key from `.env`.

```python
from dotenv import load_dotenv
load_dotenv()
```

Create the phone line to the AI (Azure, gpt-5-nano).

```python
from openai import OpenAI
from os import getenv

openai_client = OpenAI(
    base_url="https://26120057-2478-resource.openai.azure.com/openai/v1",
    api_key=getenv("AZURE_API_KEY"),
)
```

Build the assistant. Notice we give it the **database** index instead of the in-memory one, and nothing else changes.

```python
from rag_helper import RAGBase

assistant = RAGBase(sqlite_index, openai_client)
```

Ask a question.

```python
assistant.rag('I just discovered the course, can i join now')
```

**End of Lesson 9.** Our search data now lives in a file, so starting up is fast.
