# Lesson 8: RAG helper

**Simple version:** Our first notebook got long and messy. So we copied the good parts into files, like moving your best recipes from scribbled notes into a real recipe book:
- `ingest.py` downloads the FAQ cards and builds the search index.
- `rag_helper.py` has `RAGBase`, a **class**. A class is a blueprint for a robot: build it once, use it anywhere.

Now the whole RAG system takes only 5 short cells.

Import our tools from the two helper files.

```python
from rag_helper import RAGBase
from ingest import load_faq_data, build_index
```

Open the locked drawer: load the API key from `.env`.

```python
from dotenv import load_dotenv
load_dotenv()
```

Create the phone line to the AI (Azure, gpt-5-nano).

```python
from openai import OpenAI
from os import getenv

openai_client = OpenAI(
    base_url="https://26120057-2478-resource.openai.azure.com/openai/v1",
    api_key=getenv("AZURE_API_KEY"),
)
```

Download the FAQ cards, build the search index, then build our assistant robot from the `RAGBase` blueprint.

```python
documents = load_faq_data()
index = build_index(documents)

assistant = RAGBase(index, openai_client)
```

Ask the robot a question!

```python
assistant.rag('I just discovered the course, can i join now')
```
